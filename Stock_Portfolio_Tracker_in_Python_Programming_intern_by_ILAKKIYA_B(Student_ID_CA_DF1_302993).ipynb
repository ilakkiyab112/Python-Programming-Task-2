```python
import csv

# Manually defined stock prices
STOCK_PRICES = {
    "Apple": 180,
    "Tesla": 250,
    "Google": 175,
    "Microsoft": 420,
    "Amazon": 190,
    "Nvidia": 120
}


def display_stocks():
    print("\nAvailable Companies and Stock Prices")
    print("-" * 40)

    for company, price in STOCK_PRICES.items():
        print(f"{company:<15} ₹{price:,.2f}")

    print("-" * 40)


def get_portfolio():
    portfolio = {}

    while True:
        company = input(
            "\nEnter company name (or type 'done' to finish): "
        ).strip().title()

        if company == "Done":
            break

        if company not in STOCK_PRICES:
            print("Company not available. Please choose from the list.")
            continue

        try:
            quantity = int(input(f"Enter quantity of {company}: "))

            if quantity <= 0:
                print("Quantity must be greater than 0.")
                continue

        except ValueError:
            print("Please enter a valid number.")
            continue

        portfolio[company] = portfolio.get(company, 0) + quantity

    return portfolio


def calculate_portfolio(portfolio):
    details = []
    total = 0

    for company, quantity in portfolio.items():

        price = STOCK_PRICES[company]
        value = price * quantity
        total += value

        details.append({
            "company": company,
            "quantity": quantity,
            "price": price,
            "value": value
        })

    return details, total


def display_summary(details, total):

    print("\n" + "=" * 65)
    print("STOCK PORTFOLIO SUMMARY")
    print("=" * 65)

    print(
        f"{'Company':<15}"
        f"{'Quantity':<12}"
        f"{'Price':<15}"
        f"{'Investment':<15}"
    )

    print("-" * 65)

    for item in details:
        print(
            f"{item['company']:<15}"
            f"{item['quantity']:<12}"
            f"₹{item['price']:<14,.2f}"
            f"₹{item['value']:,.2f}"
        )

    print("-" * 65)
    print(f"{'TOTAL INVESTMENT':<42}₹{total:,.2f}")
    print("=" * 65)


def save_to_csv(details, total):

    filename = "stock_portfolio.csv"

    with open(filename, "w", newline="") as file:

        writer = csv.writer(file)

        writer.writerow([
            "Company",
            "Quantity",
            "Stock Price",
            "Total Investment"
        ])

        for item in details:
            writer.writerow([
                item["company"],
                item["quantity"],
                item["price"],
                item["value"]
            ])

        writer.writerow([])
        writer.writerow([
            "TOTAL INVESTMENT",
            "",
            "",
            total
        ])

    print(f"\nPortfolio saved successfully as '{filename}'.")


def main():

    print("=" * 65)
    print("              STOCK PORTFOLIO TRACKER")
    print("=" * 65)

    display_stocks()

    portfolio = get_portfolio()

    if not portfolio:
        print("\nNo companies were added to the portfolio.")
        return

    details, total = calculate_portfolio(portfolio)

    display_summary(details, total)

    save_choice = input(
        "\nDo you want to save the portfolio to a CSV file? (yes/no): "
    ).strip().lower()

    if save_choice in ("yes", "y"):
        save_to_csv(details, total)

    print("\nThank you for using the Stock Portfolio Tracker!")


if __name__ == "__main__":
    main()
```
